```python
INGESTIOIN_CONFIG = [ 
    {
         "source": "crm",
         "path": "/Volumes/databricksprash/bronze_layer_data/source_system/source_crm/sales_details.csv",
         "table": "crm_sales_details"
     }, 
     {
        "source": "crm",
        "path": "/Volumes/databricksprash/bronze_layer_data/source_system/source_crm/prd_info.csv",
        "table": "crm_prd_info"
    },
    {
         "source": "erp",
         "path": "/Volumes/databricksprash/bronze_layer_data/source_system/source_erp/CUST_AZ12.csv",
         "table": "erp_cust_az12"
    },
    {
         "source": "erp",
         "path": "/Volumes/databricksprash/bronze_layer_data/source_system/source_erp/LOC_A101.csv",
         "table": "erp_loc_a101"
    },
    {
         "source": "erp",
         "path": "/Volumes/databricksprash/bronze_layer_data/source_system/source_erp/PX_CAT_G1V2.csv",
         "table": "erp_px_cat_g1v2"
    }
]
```

Ingest bunch of files in Tables

```python
%python 

for item in INGESTIOIN_CONFIG:
    print(f"ingesting {item['source']} data from {item['path']} into table {item['table']}")
          
    df = ( 
        spark.read
            .option("header","true")
            .option("inferSchema","true")
            .csv(item['path'])
    )
    
    (df.write
        .mode("overwrite")
        .format("delta")
        .saveAsTable(f"databricksprash.bronze_layer_data.{item['table']}")
    )
```

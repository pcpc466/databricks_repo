###Transforamtion Logic 

```python
query = """ 
SELECT
cs.order_number, 
pr.porduct_key, 
cu.customer_key, 
cs.order_date, 
cs.ship_date, 
cs.due_date, 
cs.sales_amount, 
cs.quantity, 
cs.price
FROM silver_layer_data.crm_sales cs  
LEFT JOIN gold_layer_data.dim_products pr 
ON cs.product_number = pr.product_number
LEFT JOIN gold_layer_data.dim_customers cu  
ON cs.customer_id = cu.customer_id
"""
df = spark.sql(query)
display(df, 10)
```

```python
df.write.mode("overwrite").format("delta").saveAsTable("gold_layer_data.fact_sales")
```

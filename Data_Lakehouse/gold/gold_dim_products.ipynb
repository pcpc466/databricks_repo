### The Transformation Logic 

```python
query = """
SELECT 
ROW_NUMBER() OVER (ORDER BY cp.start_date, cp.product_number) AS porduct_key, -- Surrogate key 
cp.product_id, 
cp.product_number, 
cp.product_name,
cp.category_id, 
cat.category, 
cat.subcategory, 
cat.maintenance_flag,
cp.product_line, 
cp.start_date
FROM silver_layer_data.crm_products cp 
LEFT JOIN silver_layer_data.erp_product_category cat 
ON cp.category_id = cat.category_id
WHERE cp.end_date IS NULL 
LIMIT 10 -- Filter out all the historical data
"""

df = spark.sql(query)
df.limit(10).display()
```

### Writing Gold Table 

```python

df.write.mode("overwrite").format("delta").saveAsTable("databricksprash.gold_layer_data.dim_products")
```
